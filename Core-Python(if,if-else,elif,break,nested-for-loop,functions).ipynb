```python
# Conditional statements(Decisional making statements):


# It checks every time conditions, and if condition is true it will execute.
# Conditions can be taken by
# 1.comparasion operator
# 2.Logical operator
# 3.Membership operator
# 4.identity operator
# True /False (These 4 gives always Boolean)

# Block of code:
# Together they will work/ or Together they will not work…….Either all of them or none of them will execute…… Entire set is called block….....
```

```python
# IF condition

# Check condition for 1 time...
# - If statement consists of a boolean expression followed by one or more lines of code
# - Only if condition is true, it executes the lines of code inside the block
# - If the text expression is False, the statements is not executed


# if boolean condition/expression:
#    line 1
#    line 2
#    line 3
```

```python
if 1>3:
    print(True)
```

```python
if 1<3:              # since condition is true, the below whole 3 codes are executed..........
    print("True")    # part of block
    a=44             # part of block
    print(a)         # part of block
print("AI")          # Not the part of block, it is not depend on condition of above 3 codes. so just execute........
```

```python
if 1>3:              # since condition is False, the below whole 3 codes are not executed....So, entire block is ignored, not considered....
    print("True")    # part of block
    a=44             # part of block
    print(a)         # part of block
print("AI")          # Not the part of block, it is not depend on condition of above 3 codes. so just execute.. irrespective of condition.....
```

```python
if 1!=1:                    # since condition is False, the below whole 3 codes are not executed...So, entire block is ignored, not considered....
    a=54                    # part of block
    b=26                    # part of block
    print(a+b)              # part of block
print("Out of IF block")
```

```python
phone_number = input("Enter a 10-digit phone number: ").strip()

# Validate that the input is exactly 10 digits
if len(phone_number) == 10 and phone_number.isdigit():
    # Step 1: Calculate initial sum
    total_sum = sum(int(digit) for digit in phone_number)
    print("Initial sum of digits:", total_sum)

    # Step 2: Reduce to a single digit
    while total_sum >= 10:
        total_sum = sum(int(digit) for digit in str(total_sum))

    print("The final single digit sum is:", total_sum)
else:
    print("Invalid input! Please enter exactly 10 numeric digits.")
```

# For...Else

```python
# If break is not executed------------Else part will be executed
# If break is executed------------Else part will not be executed
```

```python
l=[1,2,3,4,5]      
                                               
for i in l:
    print(i)
    if i==5:                    
         break
else:
    print("loop executed without any break")
```

```python
l=[1,2,3,4]      
                                               
for i in l:             # already printed 1,,,,if 1==5, so go again to starting....2==5,
    print(i)            # again to start(****else part comes only after entire for loop completes here**or break has to be executed****
    if i==5:                    
         break
else:
    print("loop executed without any break")
```

```python
l=[1,2,3,4]      
                                               
for i in l:             # already printed 1,,,,if 1==5, so go again to starting....2==5,
    print(i)            # again to start(****else part comes only after entire for loop completes here**or break has to be executed****
    if i==4:            # here break happened at 4, so stopped at 4..*****        
         break
else:
    print("loop executed without any break")
```

```python
l=[1,2,3,4]              # IF....Else
                                               
for i in l:             # already printed 1,,,,if 1==4, if condition not worked, So else should work...***
    print(i)            # already printed 1,,,,if 2==4, if condition not worked, So else should work...***
    if i==4:            #         if 4==4 then break.
         break
    else:
     print("loop executed without any break")
```
